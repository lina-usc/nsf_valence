```python
import pandas as pd
import os
from docx import Document
```

```python
Dir_audioclip = r'X:\PROJECTS\ABC@UofSC\3-Experiments\Affective_narrative\behavioral\segmentation_audacity\output\segmented_audio'
Dir_transcript_seg = r'X:\PROJECTS\ABC@UofSC\3-Experiments\Affective_narrative\behavioral\segmentation_audacity\output\segmentation_transcript'
```

```python
List_files = [x for x in os.listdir(Dir_transcript_seg) if '.docx' in x]
dfs_segment = []
for file in List_files:
    story = file.split('_')[0]
    doc = Document(os.path.join(Dir_transcript_seg,file))
    lines = []
    for paragraph in doc.paragraphs:
        line = paragraph.text.strip()
        if line:  # Ignore empty lines
            lines.append(line)
    df = pd.DataFrame({'idx':list(range(len(lines))),'segment':lines,'story':story})
    dfs_segment.append(df)
    print(lines[1])
dfs_segment = pd.concat(dfs_segment,ignore_index=True)
dfs_segment
```
