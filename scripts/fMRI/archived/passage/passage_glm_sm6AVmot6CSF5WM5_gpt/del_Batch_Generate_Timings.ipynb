```python
import os 
import pandas as pd
import numpy as np


import nltk


#pd.set_option('display.max_rows', None)
#pd.set_option('display.max_columns', None)
```

```python
Dir_timestamps = '/work/desai-lab/ABCAI/ABC_neuroimaging-master/stimuli_and_design/timestamps/'
List_story = os.listdir(Dir_timestamps)
List_story = [file for file in List_story if '.times.' in file]
Lists_words = []
for file in List_story:

    # Using readlines()
    texts = open(os.path.join(Dir_timestamps,file), 'r')
    Lines = texts.readlines()
    List_words = []
    for line in Lines:
        if not '@ @' in line:
            List_words.append(line.replace(' \n','').replace('\n','').replace('  ',' '))
    df = pd.DataFrame(List_words)
    df.columns = ['raw']
    df[['word','onset','offset']] = df['raw'].str.split(' ',expand=True)
    df['audio'] = file.replace('.times.txt','')
    Lists_words.append(df)
dfs = pd.concat(Lists_words,ignore_index = True)

tags = nltk.pos_tag(dfs['word'])
for i,tag in enumerate(tags):
    dfs.loc[i,'PoS_tag_full'] = tag[1]
dfs
```

```python
df_scope = pd.read_csv('/work/desai-lab/xuanyang/Project/dataset/SCOPE/generateTiming/SCOPE_all.csv',encoding="latin-1")
#df_scope = pd.read_csv('E:/phd/project/Nastase/wordlevel/Passage/codes/generateTiming/SCOPE_all.csv', index_col=0,encoding="latin-1")
df_scope
```

```python
List_vars = pd.read_excel('/work/desai-lab/xuanyang/Project/Semantic/analysis/ParametricModulation/Nastase/SemanticVars/codes/SemanticVarsList.xlsx')

List_vars = list(List_vars.Variable.values)+['Valence_NRC','Arousal_NRC','Dominance_NRC']
dfs_timing = []
for i, word in zip(dfs.index,dfs.word):
    df_tmp = df_scope.loc[df_scope.Word==word,List_vars].reset_index(drop=True)
    if len(df_tmp) == 0:
        df_tmp.loc[0] = np.nan
    df_tmp_word = pd.concat([dfs.loc[i],df_tmp.loc[0]]).to_frame().T
    dfs_timing.append(df_tmp_word)

dfs_timing = pd.concat(dfs_timing,ignore_index=True)
dfs_timing
```
