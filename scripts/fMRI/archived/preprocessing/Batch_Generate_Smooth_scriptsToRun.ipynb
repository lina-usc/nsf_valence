```python
Dir_script = '/work/xy6/ABC_dataset/codes/scriptsToRun/afni-smooth'
Dir_smooth = '/work/xy6/ABC_dataset/derivatives/afni-smooth/'
Dir_fmriprep = '/work/xy6/ABC_dataset/derivatives/fmriprep/'
FWHM = 6
for subID in List_subID:
    subID = subID[4:8]
    scriptFile = os.path.join(Dir_script,'Batch_smooth_{}.sh'.format(subID))
    with open(scriptFile, "w") as f:
        f.write("#!/bin/tcsh\n")
            
    for sesID in ['visit01','visit02']:
        Dir_fmriprep_subj = os.path.join(Dir_fmriprep,'sub-'+subID,'ses-'+sesID)
        if os.path.exists(Dir_fmriprep_subj):
            print('subID:{}'.format(subID))
            Dir_smooth_subj = os.path.join(Dir_smooth,'sub-'+subID,'ses-'+sesID)
            if not os.path.exists(Dir_smooth_subj):
                os.makedirs(Dir_smooth_subj,exist_ok=True)
                os.mkdir(os.path.join(Dir_smooth_subj,'func'))
            
            
            for task in ['Fam','Passage']:
                img_fmri = os.path.join(Dir_fmriprep_subj,'func',
                                      'sub-{}_ses-{}_task-{}_space-MNI152NLin2009cAsym_res-2_desc-preproc_bold.nii.gz'.format(subID,sesID,task))
                img_mask = os.path.join(Dir_fmriprep_subj,'func',
                                      'sub-{}_ses-{}_task-{}_space-MNI152NLin2009cAsym_res-2_desc-brain_mask.nii.gz'.format(subID,sesID,task))
                
                if os.path.exists(img_fmri):
                    smoothFile = os.path.join(Dir_smooth_subj,'func',
                                      'sub-{}_ses-{}_task-{}_space-MNI152NLin2009cAsym_res-2_desc-sm6_bold.nii.gz'.format(subID,sesID,task))
                    with open(scriptFile, "a") as f:
                        f.write("echo start smoothing sub-{}-ses-{}-task-{}\n".format(subID,sesID,task))
                        f.write('/work/apps/AFNI/linux_openmp_64/3dBlurToFWHM -mask {} -FWHM {} \\\n'.format(img_mask,FWHM))
                        f.write('-input {} -prefix {} -blurmaster {} -detrend -bmall \n'.format(img_fmri,smoothFile,img_fmri))
    with open(scriptFile, "a") as f:
        f.write('echo All done. \n')
                                
                        
```
