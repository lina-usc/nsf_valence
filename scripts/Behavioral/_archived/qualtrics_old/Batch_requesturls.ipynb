```python
import requests

# Replace with your API token and data center ID
# api_token = 'YOUR_API_TOKEN'
# data_center_id = 'YOUR_DATA_CENTER_ID'

# Headers for the API requests
headers = {
    'X-API-TOKEN': api_token,
    'Content-Type': 'application/json'
}

# Function to get the list of files
def get_files():
    url = f'https://{data_center}.qualtrics.com/API/v3/libraries/UR_5uvSniYAMbURnkV/files'
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()['result']['elements']
    else:
        print(f"Error: {response.status_code}")
        return []

# Function to get the file URL
def get_file_url(file_id):
    url = f'https://{data_center}.qualtrics.com/API/v3/libraries/UR_5uvSniYAMbURnkV/files/{file_id}'
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()['result']['url']
    else:
        print(f"Error: {response.status_code}")
        return None

# Get the list of files
files = get_files()

# Retrieve URLs for each file
file_urls = []
for file in files:
    file_id = file['id']
    file_url = get_file_url(file_id)
    if file_url:
        file_urls.append(file_url)

# Save the URLs to a file
with open('file_urls.txt', 'w') as f:
    for url in file_urls:
        f.write(url + '\n')

print('File URLs have been saved to file_urls.txt')
```
