```python
import pandas as pd
import os
Dir_github = r'X:\PROJECTS\ABC@UofSC\3-Experiments\NSF_valence\github'
# Dir_segments = os.path.join(Dir_github,'derivatives','behavioral','segmentation')
Dir_labels = os.path.join(Dir_github,'derivatives','behavioral','timestamps','audacity_labels','doublechecked')
repo = 'ABC_valence'
os.listdir(Dir_labels)

dfs = []
for audio in ['Brandi_Elvis']:
    df = pd.read_csv(os.path.join(Dir_labels,f'labels_{audio}_AVR.txt'),sep='\t',header=None,names=['onset','offset','segment'])
    df['onset'] = df['onset'].astype(float)
    df['offset'] = df['offset'].astype(float)
    df['duration'] = (df['offset']-df['onset'])*1000
    df['audio'] = audio
    for i,row in df.iterrows():
        df.loc[i,'path'] = f'1.Inputs\Stimuli\{audio}\{audio}_{i+1:03}.mp3'

    dfs.append(df)
dfs = pd.concat(dfs,ignore_index=True)
dfs['Weight'] = 1
dfs['Nested'] = ''
dfs['Procedure'] = 'ProcTrial'
dfs
```

```python
subid = 1004
Dir_output = os.path.join(Dir_github,'derivatives','behavioral','Exp_physio','1.Inputs','Stimuli',f'sub-{subid}')
if not os.path.exists(Dir_output):
    os.makedirs(Dir_output)
filename = f'ListListenRateA_{subid}.txt'
dfs.loc[dfs['audio']=='Brandi_Elvis'].to_csv(os.path.join(Dir_output,filename),sep='\t',encoding='utf-8',index=False)

filename = f'ListListenRateB_{subid}.txt'
dfs.loc[dfs['audio']=='Jacob_HowCouldYou'].to_csv(os.path.join(Dir_output,filename),sep='\t',encoding='utf-8',index=False)
```
