```python
import os
import pandas as pd
import shutil
import numpy as np

Dir_github = r'X:\PROJECTS\ABC@UofSC\3-Experiments\NSF_valence\github'
repo = 'ABC_valence'
```

```python
Dir_working = r'Z:\study\phd\project\ABC\ABC'
List_files = os.listdir(Dir_working)
df_raw = pd.DataFrame(List_files,columns=['raw'])
df_raw = df_raw.loc[~df_raw['raw'].str.startswith('del_')].reset_index(drop=True)
```

```python
df = df_raw.copy()
df['raw'] = df['raw'].str.replace("\.-",'-')
df.loc[df['raw'].str.contains('.txt'),'format'] = 'txt'
df.loc[df['raw'].str.contains('.edat3'),'format'] = 'edat'
df['file'] = df['raw'].values
df['file']=df['file'].str.replace('.txt','')
df['file']=df['file'].str.replace('.edat3','')

df = df.loc[~df['file'].str.contains('Export')]
df = df.dropna()
# df['subid'] = df['raw'].str.split('-',expand=True)[1]
df[['subid_raw','sessionid']] = df['raw'].str.split('.',expand=True)[0].str.split('-',expand=True)[[1,2]]
# df['task'] = df['raw'].str.split('-',1,expand=True)[0]
df.loc[df['file'].str.contains('|'.join(['pass','Passage','Pass'])),'task'] = 'passage'
df.loc[df['file'].str.contains('|'.join(['fam','Word'])),'task'] = 'familarity'
df = df.loc[~df['sessionid'].isna()]
df_wide = df.pivot(index=['subid_raw','sessionid','task'],columns='format',values='file').reset_index()

filename = 'df_master_Eprime.csv'
df_wide.to_csv(os.path.join(Dir_github,repo,'scripts','Behavioral','check_eprime',filename),index=False)
```
