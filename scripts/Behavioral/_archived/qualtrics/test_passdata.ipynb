```python
from flask import Flask, request, jsonify
import gspread
from oauth2client.service_account import ServiceAccountCredentials

# Flask app
app = Flask(__name__)

# Google Sheets setup
scope = [
    'https://www.googleapis.com/auth/spreadsheets',
    "https://www.googleapis.com/auth/drive.file",
    "https://www.googleapis.com/auth/drive"
]
file_json = r'X:\PROJECTS\ABC@UofSC\3-Experiments\NSF_valence\github\ABC_valence\scripts\Behavioral\qualtrics\abc-narrative-69c8408d0e56.json'
creds = ServiceAccountCredentials.from_json_keyfile_name(file_json, scope)
client = gspread.authorize(creds)
sheet = client.open('SONA_exp_test').sheet1

# Flask route to check embedded data
@app.route('/check_participant', methods=['GET'])
def check_participant():
    # Get participant ID from query parameters
    participant_id = request.args.get('id')
    
    if not participant_id:
        return jsonify({"error": "Missing participant ID"}), 400
    
    # Get all data from the Google Sheet
    records = sheet.get_all_records()
    ids = [str(row['ParticipantID']) for row in records]  # Replace 'ParticipantID' with your column name
    
    # Check if the participant ID exists in the sheet
    if participant_id in ids:
        return jsonify({"status": "Yes", "message": "The participant has finished"}), 200
    else:
        return jsonify({"status": "No", "message": "The participant has not finished"}), 200

if __name__ == '__main__':
    app.run(debug=True)
```

```python
# Example Python code for searching the contact list
import requests

# Qualtrics API setup
api_token = 'your_api_token'
data_center = 'your_data_center'
list_id = 'your_contact_list_id'
participant_id = 'participant_id_from_survey'

headers = {
    'X-API-TOKEN': api_token
}

# Endpoint to fetch contacts from a list
url = f'https://{data_center}.qualtrics.com/API/v3/contactlists/{list_id}/contacts'

response = requests.get(url, headers=headers)

# Check if the participant ID is in the contact list
contacts = response.json()
contact_exists = any(contact['participant_id'] == participant_id for contact in contacts['result']['elements'])

if not contact_exists:
    # Add the participant to the contact list if not found
    add_url = f'https://{data_center}.qualtrics.com/API/v3/contactlists/{list_id}/contacts'
    add_data = {
        'firstName': 'John',  # Example: You can add custom data for the contact
        'lastName': 'Doe',
        'email': 'participant@example.com',
        'participant_id': participant_id
    }

    add_response = requests.post(add_url, json=add_data, headers=headers)
    print(f'Participant added: {add_response.status_code}')
```
