```python
def get_rating_gpt(Dir_ratings):
    List_models = [x for x in os.listdir(os.path.join(Dir_ratings,'gpt')) if 'gpt' in x]
    for model in List_models:
        List_files = os.listdir(os.path.join(Dir_ratings,'gpt',model))
        List_files = [x for x in List_files if 'rating' in x]
        for file in List_files:
            df = pd.read_csv(os.path.join(Dir_ratings,'gpt',model,file))

    return(df)
    
```
