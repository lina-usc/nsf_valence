```python
import pandas as pd
import os
import platform
import json
# from docx import Document
import numpy as np
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
import matplotlib.pyplot as plt
import seaborn as sns

if platform.system() == 'Linux':
    Dir_github = r'/work/xy6/Project/ABC/github/' # HPC
else:
    Dir_github = r'X:\PROJECTS\ABC@UofSC\3-Experiments\NSF_valence\github' # Local computer
repo = 'ABC_valence'
pipeline = 'makeTimings'
Dir_working = os.path.join(Dir_github,repo,'scripts',pipeline)
Dir_ratings = os.path.join(Dir_github,'derivatives','behavioral','ratings')
Dir_output = os.path.join(Dir_github,'derivatives','behavioral','timings')
if not os.path.exists(Dir_output):
    os.makedirs(Dir_output)
```

```python
model = 'gpt-4o-2024-08-06'
df_ratings = pd.read_csv(os.path.join(Dir_ratings,'gpt',model,f"rating_gpt_{model}.csv"))
dfs_timings = df_ratings[['story','idx_raw','segment','onset','offset','V','A']].groupby(['story','idx_raw','segment']).mean().reset_index()
dfs_timings['rater'] = model
for story in dfs_timings['story'].unique():
    df = dfs_timings.loc[dfs_timings['story']==story]
    Dir_output2 = os.path.join(Dir_output,model)
    if not os.path.exists(Dir_output2):
        os.makedirs(Dir_output2)
    df.to_csv(os.path.join(Dir_output2,f'timing_{model}_{story}_bySeg.csv'))
```
