# Make the CSF mask to visualize the preprocessing quality

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import json
import subprocess
Dir_github = '/work/desai-lab/xuanyang/Project/ABC/github'
Dir_dataset = '/work/desai-lab/ABC_dataset'
Dir_preprocess = os.path.join(Dir_dataset,'derivatives','fmriprep')
project = 'ABC_valence'
```

```python
df_master = pd.read_csv(os.path.join(Dir_github,'derivatives','test_Fam','df_master_Fam120.csv'))
df_master
```

```python
List_masks = []
for i, row in df_master.iterrows():
    subid = int(row['subid'])
    visit = int(row['visit'])
    Dir_subject = os.path.join(Dir_preprocess,f"sub-{subid}",f"ses-visit0{visit}")

    # T1 mask
    mask = os.path.join(Dir_subject,'anat',f'sub-{subid}_ses-visit0{visit}_space-MNI152NLin2009cAsym_res-2_desc-brain_mask.nii.gz')
    if not os.path.exists(mask):
        mask = os.path.join(Dir_preprocess,f"sub-{subid}",'anat',f'sub-{subid}_space-MNI152NLin2009cAsym_res-2_label-CSF_probseg.nii.gz')
        if not os.path.exists(mask):
            print(subid)
    List_masks.append(mask)
```

```python


Dir_output = os.path.join(Dir_github,project,'masks')
os.chdir(Dir_output)

file = 'img_CSFprob_Fam120.nii.gz'
command = f"/work/apps/AFNI/linux_openmp_64/3dMean -prefix {file} {' '.join(List_masks)}"
subprocess.run(command,shell=True)

threshold = 99
img_mask1 = f'mask_CSFprob{threshold}_Fam120.nii.gz'
command = f"/work/apps/AFNI/linux_openmp_64/3dcalc -a {file} -expr 'step(a-0.{threshold})' -prefix {img_mask1}"
subprocess.run(command,shell=True)
```
