```python
import os
import pandas as pd
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
```

The ABC dataset originally has 23 stories x 2 voices = 46 audio clips (see https://github.com/NickRiccardi/ABC_neuroimaging)

```python
Dir_audio = r'X:\PROJECTS\ABC@UofSC\3-Experiments\Affective_narrative\Audio_normalized\Normalized\English_language'
df_master = pd.DataFrame({'audio':os.listdir(Dir_audio)})
df_master[['voice','story']] = df_master['audio'].str.split('_',1,expand=True)
df_master['story'] = df_master.story.str.replace('.wav','')
df_master.head()
```

The story transcripts were segmented by two RAs

```python
Dir_SegLabels = r'X:\PROJECTS\ABC@UofSC\3-Experiments\Affective_narrative\behavioral\segmentation_audacity\output\segmentation_labels'
Dir_transcript = r'X:\PROJECTS\ABC@UofSC\3-Experiments\Affective_narrative\behavioral\segmentation_audacity\output\segmentation_transcript'
for i, row in df_master.iterrows():
    # Segment Labels
    filename = os.path.join(Dir_SegLabels,f"{row['voice']}_{row['story']}_labels.txt")
    if os.path.exists(filename):
        df_master.loc[i,'SegLabel'] = 'Done'

    # Transcripts
    filename = os.path.join(Dir_transcript,f"{row['story']}_transcript_finished.docx")
    if os.path.exists(filename):
        df_master.loc[i,'SegTranscript'] = 'Done'
df_master.loc[df_master['SegTranscript'].isna()]
```
