```python
import os
import pandas as pd
import numpy as np
```

```python
Dir_github = r'X:\PROJECTS\ABC@UofSC\3-Experiments\NSF_valence\github\nsf_valence'
Dir_ratings = os.path.join(Dir_github,'derivatives','behavioral','ratings','audio')
df_ratings = pd.read_csv(os.path.join(Dir_ratings,'ratings_audio_att2_comp2.csv'), dtype={'idx': str,'ethnicity':str},keep_default_na=False)
```

```python
Dir_segmentations = os.path.join(Dir_github,'derivatives','behavioral','segmentation')

List_files = [x for x in os.listdir(Dir_segmentations) if x.startswith('segmentation_')]
df_segmentations = []
for file in List_files:
    df = pd.read_csv(os.path.join(Dir_segmentations,file))
    df_segmentations.append(df)

    
df_segmentations = pd.concat(df_segmentations,ignore_index=True)
df_segmentations
```

```python
Dir_timestamps = os.path.join(Dir_github,'derivatives','behavioral','timestamps_segment')
List_files = [x for x in os.listdir(Dir_timestamps) if x.startswith('timestamps_')]
df_timestamps = []
for file in List_files:
    df = pd.read_csv(os.path.join(Dir_timestamps,file))
    df_timestamps.append(df)
df_timestamps = pd.concat(df_timestamps,ignore_index=True)
df_timestamps['duration'] = df_timestamps['offset'] - df_timestamps['onset']
```

```python
df_summary = df_segmentations.groupby(['narrative']).max()[['sentID','segID']]
df_summary['Nwords'] = df_segmentations.groupby(['narrative']).sum()['Nwords']
df_summary['words/segment'] = df_summary['Nwords']/df_summary['segID']
df_summary.columns = ['Total sentences','Total segments','Total words','Words/segment']
df_summary['Words/segment'] = df_summary['Words/segment'].round(2)

df_summary['Seconds/segment'] = df_timestamps.groupby(['narrative']).mean(numeric_only=True)['duration'].round(2)
df_summary.loc['Average',['Total sentences','Total segments','Total words','Words/segment','Seconds/segment']] = df_summary.mean().round(2)
Dir_output = os.path.join(Dir_github,'derivatives','behavioral','figures&tables')
os.makedirs(Dir_output,exist_ok=True)
df_summary.to_csv(os.path.join(Dir_output,'table_segmentation_summary.csv'))
```
