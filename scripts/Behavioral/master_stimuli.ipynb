This script is used to make a master sheet that contains the correct names of all narratives and audio recordings

```python
import os
import pandas as pd
Dir_github = r'X:\PROJECTS\ABC@UofSC\3-Experiments\NSF_valence\github\nsf_valence'
Dir_transcript = os.path.join(Dir_github,'stimuli','transcripts')
df_transcript = pd.DataFrame({'narrative':[x for x in os.listdir(Dir_transcript) if '.docx' in x]})
df_transcript['narrative'] = df_transcript['narrative'].str.replace('.docx','')
```

```python
Dir_audio = os.path.join(Dir_github,'stimuli','audios','English_language')
df_audio = pd.DataFrame({'voice_narrative':[x for x in os.listdir(Dir_audio) if '.wav' in x]})
df_audio['voice_narrative'] = df_audio['voice_narrative'].str.replace('.wav','')
df_audio[['voice','narrative']] = df_audio['voice_narrative'].str.split('_',expand=True)
List_voice_F = ['Allison', 'Brandi', 'Jewell','WavenetF']
List_voice_M = ['Jack', 'Jacob', 'WavenetB']
df_audio.loc[df_audio['voice'].isin(List_voice_F),'voice_sex'] = 'female'
df_audio.loc[df_audio['voice'].isin(List_voice_M),'voice_sex'] = 'male'
df_audio = df_audio.sort_values(['narrative','voice_sex']).reset_index(drop=True)
set(df_transcript['narrative'].unique()) == set(df_audio['narrative'].unique())
```

```python
filename = os.path.join(Dir_github,'stimuli','master_stimuli.csv')
df_audio.to_csv(filename,index=False)
print(f"The master sheet of audio stimuli was exported to {filename}")
```
